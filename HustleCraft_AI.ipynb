```python
# Install required libraries
!pip install -q gradio huggingface_hub

import os
import gradio as gr
from huggingface_hub import InferenceClient

# Initialize the Hugging Face Inference Client.
# This uses the completely free serverless API endpoint.
# You can optionally set a HF_TOKEN in your environment or Colab secrets,
# but Qwen2.5-7B-Instruct often works directly or with a free community token.
client = InferenceClient(model="Qwen/Qwen2.5-7B-Instruct")

def generate_hustle_content(raw_notes, task_type, tone):
    if not raw_notes.strip():
        return "Please enter some raw notes to get started!"

    # Craft the system prompt based on the chosen side hustle task
    if task_type == "Upwork Proposal":
        system_prompt = (
            "You are an expert freelancer. Write a highly persuasive, professional, "
            "and customized Upwork proposal based on the client's job requirements. "
            "Highlight relevant skills, ask clarifying questions, and keep it engaging."
        )
    elif task_type == "Product Description":
        system_prompt = (
            "You are a brilliant e-commerce copywriter. Write an enticing, SEO-friendly "
            "product description based on the provided notes. Emphasize benefits and key features."
        )
    else:  # Invoice Notes & Thank You
        system_prompt = (
            "You are a professional business owner. Write a polite, detailed, "
            "and professional invoice note and thank-you message to a client based on the completed work details."
        )

    # Build the full instruction
    prompt = f"Tone: {tone}\nRaw Notes:\n{raw_notes}\n\nGenerate the final polished output based on your role:"

    try:
        # Request completion from the free Hugging Face serverless API
        response = client.chat_completion(
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": prompt}
            ],
            max_tokens=1000,
            temperature=0.7
        )
        return response.choices[0].message.content
    except Exception as e:
        # Friendly fallback explanation if rate limits are hit
        return (
            f"An error occurred or the free serverless API rate limit was hit: {str(e)}\n\n"
            "Tip: If you have a free Hugging Face account, you can add your free API token "
            "by running `os.environ['HF_TOKEN'] = 'your_token'` before executing the app for uninterrupted access."
        )

# Design the HustleCraft AI Gradio App
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🛠️ HustleCraft AI")
    gr.Markdown(
        "### Hacktoberfest Week 2: Side Hustle Edition\n"
        "Turn raw notes into polished Upwork proposals, product descriptions, or professional invoice messages "
        "instantly using the free Hugging Face Serverless API."
    )

    with gr.Row():
        with gr.Column():
            raw_notes = gr.Textbox(
                label="Raw Job / Product Notes",
                placeholder="e.g., Build a simple React landing page for a coffee shop, needs to be done in 3 days, clean style...",
                lines=6
            )
            task_type = gr.Dropdown(
                choices=["Upwork Proposal", "Product Description", "Invoice Notes & Thank You"],
                value="Upwork Proposal",
                label="Target Output"
            )
            tone = gr.Radio(
                choices=["Professional", "Persuasive", "Casual", "Enthusiastic"],
                value="Professional",
                label="Writing Tone"
            )
            generate_btn = gr.Button("Craft Content", variant="primary")

        with gr.Column():
            output = gr.Markdown(label="Generated Hustle Output")

    generate_btn.click(
        fn=generate_hustle_content,
        inputs=[raw_notes, task_type, tone],
        outputs=output
    )

# Launch the Gradio app with a public share link!
demo.launch(share=True, debug=True)
```
